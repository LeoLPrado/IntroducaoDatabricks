# Basic
- Creating a DataFrame
- Selecting columns
- Filtering rows
- Adding Columns
- Grouping and Aggregating
- Dropping Columns
- Distinct Values


```python

data = [("Alice", 34, "NYC", "F"), ("Bob", 45, "LA", "M"), ("Leo", 19, "NYC", "M")]
```

```python
# 1
df = spark.createDataFrame(data, ["name", "age", "city", "gender"])
display(df)
```

```python
# 2 
df_selecionado = df.select('name', 'city')
display(df)
```

```python
# 3
from pyspark.sql import functions as F

df_filtrado = df.filter(F.col('age') > 30) #ou df_filtado = df.filter(df.age > 30)
display(df_filtrado)
```

```python
# 4
df_new = df.withColumn('age_in_months', df.age * 12)
display(df_new)
```

```python
# 5
data = [
    ("Sales", "Alice", 5000),
    ("Sales", "Bob", 4500),
    ("Engineering", "Charlie", 6000),
    ("Engineering", "Diana", 6500),
    ("Engineering", "Eve", 5500)
]
df = spark.createDataFrame(data, ["department", "name", "salary"])
display(df)
```

```python
# 6
from pyspark.sql import functions as F

df_filtrado = (
    df.groupBy("department")
    .agg(
        F.round(F.avg('salary'), 2).alias('avg_salary')
    )
)
display(df_filtrado)
```

```python
# 7

dropped_df = df.drop('department')
display(dropped_df)
```

```python
# 8
data = [("Alice", "NYC"), ("Bob", "LA"), ("Alice", "NYC"), ("Charlie", "NYC")]
df = spark.createDataFrame(data, ["name", "city"])

df_distinct = df.distinct()
display(df_distinct)

df_distinct2 = df.select("city").distinct()
display(df_distinct2)
```

# Columns Operations
- Renaming Columns
- Literal Values
- Conditional Logic with when/otherwise
- Type Casting


```python
data = [("Alice", 34), ("Bob", 45)]
df = spark.createDataFrame(data, ["name", "age"])

df_renomeado = df.withColumnRenamed('age', 'idade')
df_renomeado2 = df.withColumnsRenamed({"age": "idade", "name": "nome"})
display(df_renomeado)
display(df_renomeado2)
```

```python
from pyspark.sql.functions import lit

df_local = df.withColumn('country', lit('Brazil'))
display(df_local)
```

```python
from pyspark.sql.functions import when, col

df_agrupado = df.withColumn(
    'faixa_etaria',
    when(col('age') < 30, 'Young')
    .when((col("age") > 30) & (col("age") < 60), "Adult")
    .otherwise('Senior')
)

display(df_agrupado)
```

```python
from pyspark.sql.functions import col

df_cast = df.withColumn('age_correct', col('age').cast('integer'))

display(df_cast)
```

# Strings Functions
  - Case
  - Concatenation
  - Substring and Length

```python
from pyspark.sql.functions import col, upper, lower, initcap

data = [("alice smith",), ("BOB JONES",), ("Charlie Brown",)]
df = spark.createDataFrame(data, ["name"])

df_upper = df.withColumn('name_upper', upper(col('name'))) 
display(df_upper)
```

```python
df_lower = df.withColumn('name_lower', lower(col('name'))) 
display(df_lower)
```

```python
df_initcap = df.withColumn('name_initcap', initcap(col('name'))) 
display(df_initcap)
```

```python
from pyspark.sql.functions import concat, lit, length, col

data = [("Alice", "Smith"), ("Bob", "Jones")]
df = spark.createDataFrame(data, ["first", "last"])

df_nome_completo = df.withColumn('full_name', concat(col('first'), lit(' '), col('last')))
display(df_nome_completo)
```

```python
data = [("Alice",), ("Bob",), ("Charlotte",)]
df = spark.createDataFrame(data, ["name"])

df_len = df.withColumn('name_len', length('name'))
display(df_len)
```
